Dans un terminal, lancer `uv run python sav_server.py`. Configurer `GROQ_API_KEY` dans `.env` et choisir le noyau `.venv`.

Les deux exemples utilisent les donnees fictives du PX-200.

```python
from sav_graph import graph

examples = [
    (
        "Mon imprimante PX-200 affiche un voyant clignotant. "
        "Que signifie le pictogramme sur la photo jointe ?",
        "img_px200_01",
    ),
    (
        "Est-ce que mon imprimante est encore sous garantie ? "
        "Son numero de serie est PX200-0042.",
        "",
    ),
]

for question, image_ref in examples:
    print(question)
    async for event in graph.astream(
        {"question": question, "image_ref": image_ref}, stream_mode="updates"
    ):
        print(event)
    print()
```
